```python
# Cell 1: Setup and Imports
from ultralytics import YOLO
import torch
import yaml
import os

# Verify GPU
print(f"CUDA Available: {torch.cuda.is_available()}")
print(f"Device: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")
```

```python
# Cell 2: Dataset Path Setup
import os

# Change to project root
os.chdir('..')
print(f"Current directory: {os.getcwd()}")

# Set dataset path
DATASET_PATH = "data/merged_dataset/data.yaml"

# Verify dataset exists
if os.path.exists(DATASET_PATH):
    print(f"✅ Dataset found: {DATASET_PATH}")
else:
    print(f"❌ Dataset NOT found: {DATASET_PATH}")
```

```python
# Cell 3: Inspect Dataset
with open(DATASET_PATH, 'r') as f:
    data_config = yaml.safe_load(f)
    
print("Dataset Configuration:")
print(f"Classes: {data_config['names']}")
print(f"Number of classes: {data_config['nc']}")
print(f"Train path: {data_config['train']}")
print(f"Val path: {data_config['val']}")
```

```python
# Cell 4: Load Pre-trained Model
model = YOLO('yolov8n.pt')  # Nano model - fast and efficient
print("✅ YOLOv8 Nano model loaded")
```

```python
# Cell 5: Start Training
results = model.train(
    data=DATASET_PATH,
    epochs=50,                    # 50 epochs
    imgsz=640,                    # Image size
    batch=16,                     # Batch size for your 6GB GPU
    device=0,                     # Use GPU
    project='models',
    name='leopard_v1',
    patience=10,                  # Early stopping
    save=True,
    plots=True,
    cache=True,                   # Cache images in RAM
    workers=4,
    verbose=True
)

print("\n✅ Training complete!")
```

```python
# Cell 6: Validate Model
metrics = model.val()
print(f"\n📊 Validation Results:")
print(f"mAP50: {metrics.box.map50:.3f}")
print(f"mAP50-95: {metrics.box.map:.3f}")
```

```python
# Cell 7: Test on Sample Image
import matplotlib.pyplot as plt
from PIL import Image

# Test with an image from validation set
test_img_path = "data/merged_dataset/valid/images"

# Get first image from validation
import glob
test_images = glob.glob(f"{test_img_path}/*.jpg")[:3]

for img_path in test_images:
    results = model(img_path)
    results[0].show()  # Display result
    print(f"Tested: {img_path}")
```

```python
# Cell 8: Save Best Model
import shutil
import os

# Copy best model to easy location
source = 'models/leopard_v12/weights/best.pt'
destination = 'models/best.pt'

if os.path.exists(source):
    shutil.copy(source, destination)
    print(f"✅ Best model saved to: {destination}")
else:
    print(f"❌ Model not found at: {source}")
```
